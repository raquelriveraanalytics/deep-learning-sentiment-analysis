## Introduction: IMDb Sentiment Classification Task

In this assignment, we will use the **IMDb movie reviews dataset**, a widely-used benchmark for text classification. The dataset contains **50,000 movie reviews** from IMDb, labeled as either **positive (1)** or **negative (0)** based on sentiment.

The main objective is to **build and compare different models for sentiment classification**:

- **CNN-only model**: Captures local patterns in text (e.g., n-grams).
- **LSTM-only model**: Captures long-term dependencies and word order.
- **CNN-LSTM model**: Combines local and sequential modeling.
- **SVM with TF-IDF**: A traditional machine learning benchmark for text classification.

This task helps illustrate the strengths and weaknesses of deep learning models compared to classical approaches. You will:
- Preprocess the data (tokenization and padding),
- Train each model,
- Evaluate performance using test accuracy,
- Discuss trade-offs between model complexity, training time, and interpretability.


```python
# Step 1: Imports
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, MaxPooling1D, GlobalMaxPooling1D, LSTM, Dense, Dropout

# Step 2: Load IMDb dataset
vocab_size = 10000
max_len = 500

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=vocab_size)
x_train = pad_sequences(x_train, maxlen=max_len, padding='post', truncating='post')
x_test = pad_sequences(x_test, maxlen=max_len, padding='post', truncating='post')
```

```python
# Build CNN + LSTM model
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=128, input_length=max_len),
    Conv1D(filters=32, kernel_size=5, activation='relu'),
    MaxPooling1D(pool_size=2),
    LSTM(32),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

# Step 4: Compile model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

# Step 5: Train model
model.fit(x_train, y_train, epochs=4, batch_size=128, validation_data=(x_test, y_test))
```

```python
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"\n CNN+LSTM Test Accuracy: {accuracy:.4f}")
```

```python
# Step 1: Import Libraries
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.datasets.imdb import get_word_index

# Step 2: Load raw IMDb data (text version not available by default)
# We'll decode the word indices back into text
vocab_size = 10000
max_len = 500

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=vocab_size)
word_index = get_word_index()
reverse_word_index = {value: key for key, value in word_index.items()}

def decode_review(encoded):
    return " ".join([reverse_word_index.get(i - 3, '?') for i in encoded])

x_train_text = [decode_review(x) for x in x_train]
x_test_text = [decode_review(x) for x in x_test]

# Step 3: Define SVM pipeline with TF-IDF
svm_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=10000, stop_words='english')),
    ('svm', LinearSVC())
])

# Step 4: Train and evaluate
svm_pipeline.fit(x_train_text, y_train)
y_pred = svm_pipeline.predict(x_test_text)
svm_acc = accuracy_score(y_test, y_pred)

print(f"SVM Test Accuracy: {svm_acc:.4f}")
```

##  Model Performance Summary

| **SVM (TF-IDF)** | 0.8594      |

### Obervations

-  **CNN** outperformed all other models with an accuracy of **88.5%**, effectively capturing local n-gram patterns that are useful for sentiment classification.
-  **LSTM** and **CNN + LSTM** models performed poorly (around **51% accuracy**, close to random guessing), likely due to overfitting, suboptimal training, or insufficient epochs.
- **SVM with TF-IDF** performed well (**85.9% accuracy**) despite being a classical machine learning method, demonstrating the strength of simple, interpretable pipelines for text classification tasks.

### Takeaway

While deep learning models like LSTM and CNN+LSTM are powerful, they may require more careful tuning (e.g., learning rate, regularization, or more training epochs) to outperform simpler models. In this case, **CNN and SVM** provided strong baseline performance.
